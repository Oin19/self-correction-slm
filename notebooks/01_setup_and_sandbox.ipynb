## Step 0: Repository Setup (Kaggle / Colab)
Clones the project repository into the notebook environment and registers `src/` on the Python path. Skip this cell if running locally with the repo already cloned.

```python
import os, sys, subprocess

# --- Configuration ---
GITHUB_TOKEN = ""  # Paste your GitHub PAT here if the repo is private
REPO_OWNER   = "Oin19"
REPO_NAME    = "self-correction-llm-rl"
REPO_DIR     = f"/kaggle/working/{REPO_NAME}"

# --- Clone if not already present ---
if not os.path.exists(os.path.join(REPO_DIR, "src")):
    if GITHUB_TOKEN:
        repo_url = f"https://{GITHUB_TOKEN}@github.com/{REPO_OWNER}/{REPO_NAME}.git"
    else:
        repo_url = f"https://github.com/{REPO_OWNER}/{REPO_NAME}.git"
    print("Cloning repository from GitHub...")
    result = subprocess.run(["git", "clone", "-q", repo_url, REPO_DIR], capture_output=True, text=True)
    if result.returncode != 0:
        print("Clone failed:", result.stderr)
    else:
        print("Repository cloned successfully.")
else:
    print(f"Repository already present at {REPO_DIR}")

# --- Register src/ on Python path ---
for candidate in [REPO_DIR, os.path.abspath("."), os.path.abspath("..")]:
    if os.path.exists(os.path.join(candidate, "src")) and candidate not in sys.path:
        sys.path.insert(0, candidate)
        print(f"Added to sys.path: {candidate}")
        break

print("Setup complete.")
```

## Step 2: Python Execution Sandbox (Verification Checks V3 & V4)
Runs untrusted Python code in isolated subprocesses with strict timeout (8s) and memory caps (1GB). Classifies execution status into:
- `AC`: All Correct
- `WA`: Wrong Answer
- `TLE`: Time Limit Exceeded
- `MLE`: Memory Limit Exceeded
- `CE`: Compilation / Syntax Error
- `RE`: Runtime Error (Crash)

```python
from src.execution.executor import run_code
from src.execution.status import STATUSES, STATUS_DESCRIPTIONS

print("Available Execution Statuses:")
for s, desc in STATUS_DESCRIPTIONS.items():
    print(f" - {s}: {desc}")

# Test 1: All Correct (AC)
res_ac = run_code("print(2 + 2)")
print(f"\nTest 1 (AC): {res_ac}")
assert res_ac["status"] == "AC"

# Test 2: Syntax Error (CE)
res_ce = run_code("def foo():")
print(f"Test 2 (CE): {res_ce}")
assert res_ce["status"] == "CE"

# Test 3: Runtime Exception (RE)
res_re = run_code("print(1 / 0)")
print(f"Test 3 (RE): {res_re}")
assert res_re["status"] == "RE"

# Test 4: Time Limit Exceeded (TLE) — reduced timeout to 0.5s for speed
res_tle = run_code("import time; time.sleep(2)", timeout=0.5)
print(f"Test 4 (TLE): {res_tle}")
assert res_tle["status"] == "TLE"

print("\nVerification Checks V3 & V4 Passed!")
```
